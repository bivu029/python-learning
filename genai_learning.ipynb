```python
def get_high_earner(employees):
    high_earner=[]
    for emp in employees:
        if emp["salary"]>50000:
            emp_name_salary=(emp["name"],emp["salary"])
            high_earner.append(emp_name_salary)
    return high_earner
```

```python
employees = [
    {"name": "A", "salary": 60000},
    {"name": "B", "salary": 45000},
    {"name": "C", "salary": 75000},
    {"name": "D", "salary": 55000}
]
earner= get_high_earner(employees)
print(earner)
```

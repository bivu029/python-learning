```python
def get_high_earner(employees):
    high_earner=[]
    for emp in employees:
        if emp["salary"]>50000:
            emp_name_salary=(emp["name"],emp["salary"])
            high_earner.append(emp_name_salary)
    return high_earner
```

```python
import json
person={
    "name":"bivas",
    "role": "Data Analyst",
    "skills": ["Python", "SQL", "Microsoft Fabric"]
}
json_data=json.dumps(person)
print(json_data)
print(type(json_data))
```

Json to dictionary

```python
json_string='''
{
    "name": "Bivas",
    "role": "Data Analyst",
    "skills": ["Python", "SQL", "Microsoft Fabric"]
}
'''
person=json.loads(json_string)
print(person)
print(type(person))
```

a header is metadata we send during api call which contain information like authentication much more

```python
import requests
headers={
    "Authorization": "Bearer test-api-key",
    "X-Developer": "Bivas"
}

response=requests.get( "https://jsonplaceholder.typicode.com/users/1",headers=headers)
print(response.status_code)
print(response.json())
```

```python
import requests
import pandas as pd

response= requests.get("https://jsonplaceholder.typicode.com/users")

users= response.json()

df= pd.DataFrame(users)
print(df)
```

```python
address_df=pd.json_normalize(users)
print(address_df.columns.tolist())
print(address_df.head())
```

```python
clean_df = address_df[
    [
        "id",
        "name",
        "email",
        "address.city",
        "address.geo.lat",
        "address.geo.lng",
        "company.name"
    ]
]

print(clean_df)
```

```python
clean_df=clean_df.rename(columns={

    "address.city": "city",
    "address.geo.lat": "latitude",
    "address.geo.lng": "longitude",
    "company.name": "company"
})

print(clean_df.head())
```

```python
a = {
    "name": "Bivas",
    "skills": ["Python", "SQL"]
}

b = a.copy()

b["skills"].append("Fabric")

print(a)
print(b)
```

```python
#deep copy 
import copy
a = {
    "name": "Bivas",
    "skills": ["Python", "SQL"]
}
b = copy.deepcopy(a)
b["skills"].append("Fabric")
print(a)
print(b)
print(a is b)
print(a["skills"] is b["skills"])
```

*ARGS


```python
def add_numbers(*args):
    print(*args)

add_numbers(1,2,3)
```

```python
from ast import arg


def addition_num(*args):
    total_sum=0
    for i in args:
        total_sum +=i
    return total_sum
s=addition_num(1,2,3,4,5,6,7)
print(s)
```

```python
def show_employees(**kwargs):
    print(kwargs)

show_employees(
    name="Bivas",
    role="Data Engineer",
    experience=4.8
)
```

```python
def test(*args,**kwargs):
    print(args)
    print(kwargs)

test(
    10,
    20,
    name="Bivas",
    role="Data Engineer"
)
```

```python
def square(x):
    print(x*x)
square(2)
```

```python
square =lambda x:x*x

print(square(2))
```

```python
employees = [
    {"name": "A", "salary": 60000},
    {"name": "B", "salary": 45000},
    {"name": "C", "salary": 75000}
]

sorted_emp= sorted(employees,key=lambda emp:emp["salary"])
print(sorted_emp)
```
